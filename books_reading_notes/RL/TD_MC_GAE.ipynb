- $G_t = R_{t+1} + γR_{t+2} + γ²R_{t+3} + ...$
- $V^{\pi_\theta}(s)= E_\pi[G_t | S_t=s] $
- $Q^{\pi_\theta}(s_t, a_t) = \mathbb{E}_{\pi_\theta} [G_t | S_t=s_t, A_t=a_t] $
- mc vs. td
    - mc: $V(S_t) ← V(S_t) + α [G_t - V(S_t)]$
        - $V^{\pi_\theta}(s)= E_\pi[G_t | S_t=s] $
        - 只要你采样的回合足够多，$G_t$ 的平均值会收敛到真实的价值 $v_π(s)$。因此，它是无偏的。
    - td: $V(S_t) ← V(S_t) + α [ (R_{t+1} + γV(S_{t+1})) - V(S_t) ]$
        - TD target: $R_{t+1} + γV(S_{t+1})$
        - 在学习的初期，$V(S_{t+1})$ 的估计几乎肯定是错的，而且错得可能很离谱。你用一个不准确的估计值来更新当前的估计值，这个过程就引入了 偏差。这个过程也叫做 自举 (Bootstrapping)，即用后续的估计值来更新当前的估计值。

$$
\delta_t^V = R_{t+1} + \gamma V(S_{t+1}) - V(S_t)
$$

$$
\begin{align*}
\hat{A}_t^{\text{GAE}(\gamma, \lambda)} &= \sum_{l=0}^{\infty} (\gamma \lambda)^l \delta_{t+l}^V \\
&= \delta_t^V + (\gamma \lambda) \delta_{t+1}^V + (\gamma \lambda)^2 \delta_{t+2}^V + \dots
\end{align*}
$$

- 当 $\lambda = 0$:
    - $\hat{A}_t^{\text{GAE}(\gamma, 0)} = \delta_t^V = R_{t+1} + \gamma V(S_{t+1}) - V(S_t)$
    - 这是基于 One-Step TD 的优势估计（高偏差，低方差）。
- 当 $\lambda = 1$:
    - $\hat{A}_t^{\text{GAE}(\gamma, 1)} = \sum_{l=0}^{\infty} \gamma^l \delta_{t+l}^V = \sum_{k=0}^{\infty} \gamma^k R_{t+k+1} - V(S_t)$
    - 这是基于 蒙特卡洛 的优势估计（低偏差，高方差）。
