### concepts

- bellman updates
    - $V(s) \leftarrow V(s) + \alpha \left[ R + \gamma V(s') - V(s) \right]$
    - $Q(s, a) \leftarrow Q(s, a) + \alpha \left[ \underbrace{r + \gamma \max_{a'} Q(s', a')}_{\text{TD Target for Q}} - Q(s, a) \right]$
        - $Q_T(\mathbf{s}, \mathbf{a}) = r + \max_{\mathbf{a}'} Q_{\theta}(\mathbf{s}', \mathbf{a}')$
            - 严谨点：TD Target = $r + \gamma \max_{a'} Q(s', a')$
