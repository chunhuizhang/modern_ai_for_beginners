- $Q(s_t,a_t)=\sum_{t'=t}^T\mathbb E_{\pi_\theta}[r(s_{t'},a_{t'})|s_t,a_t]$
    - true expected reward to go
    - $Q^\pi(s, a) = \mathbb{E}_{a\sim \pi} [G_t \vert S_t = s, A_t = a]$
- $V(s_t)=\mathbb E_{a_t\sim \pi_\theta(a_t|s_t)}[Q(s_t,a_t)]$ 作为降方差要减去的 baseline（基线的最佳选择是 value function）
    - $V^\pi (s) = \mathbb{E}_{a\sim \pi} [G_t \vert S_t = s]$
- $A(s_t,a_t)=Q(s_t,a_t) - V(s_t)$: advantage function

### $V(s_t)$

- $V^\pi(s) = \mathbb{E}_\pi \left[ \sum_{k=0}^{\infty} \gamma^k r_{t+k+1} \mid s_t = s \right]$
- $Q^\pi(s, a) = \mathbb {E}_\pi \left[ \sum_{k=0}^{\infty} \gamma^k r_{t+k+1} \mid s_t = s, a_t = a \right]$
- 对于一个样本 $(s_t,a_t,r_t,s_{t+1})$
    - 损失函数定义为TD误差的平方：$\mathcal L_{w}=\delta^2=(r_t+\gamma V_w(s_{t+1})-V_w(s_t))^2$
    - 在实际训练中，我们会从经验中采样一批数据，计算这批数据的平均损失：$\mathcal L_{w}=\mathbb E_{(s,r,s')\sim \mathcal D}(r+\gamma V_w(s')-V_w(s))^2$
