$$
p(y|x)=\sum_z p(z|x)p(y|x,z)
$$

- $p(y|x)=\sum_zp(y,z|x)$
    - $p(y,z|x)=p(z|x)p(y|x,z)$（等式两边同时乘以 $p(x)$）
    - $p(A, B | C) = p(A | B, C) \cdot p(B | C)$
- MoE 的视角
    - $p(z|x)$：你还有一个“调度员”（Gating Network）。拿到任务 $x$，调度员会判断这个任务更适合哪个专家来处理，并给每个专家一个权重或概率 $p(z|x)$
    - $p(y|x,z)$：专家网络，每个专家 $z$ 都会根据任务 $x$ 给出自己的预测 $p(y|x,z)$
    - $p_M(y|x)$: 最终的预测结果，是所有专家的预测结果的加权平均。权重就是调度员分配的权重。这样，模型就综合了所有专家的智慧，得出了一个更鲁棒的最终预测。
        - M 刻画的是 Marginal：给定输入 $x$ 的条件下，输出为 $y$ 的边缘概率
