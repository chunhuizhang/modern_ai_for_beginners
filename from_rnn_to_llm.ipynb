### n-gram language model 的起源

- Shannon (1948) 2-gram
- 3-gram
    - 将文本语料库 $C$ 预处理成 token 序列：$C=(w_1, w_2, \cdots, w_T)$
    - 概率估计：对于 trigram 模型，我们需要估计给定前两个词 $(w_{i-2},w_{i-1})$ 的情况下，下一个单词 $w_i$ 的概率，即 $P(w_i|w_{i-2}, w_{i-1})$
        - 这个条件概率可以通过最大似然估计 (Maximum Likelihood Estimation, MLE) 从语料库中计算得出。
        - $P(w_i|w_{i-2}, w_{i-1})=\frac{\text{Count}(w_{i-2},w_{i-1},w_i)}{\text{Count}(w_{i-2},w_{i-1})}$ 

### RNN => LSTM => Attention (Transformer)

> Use RNN to model text.

- RNN(2010),
    - $h_t=f(W_{hh}h_{t-1}+W_{xh}x_t+b_h)$
    - 在时间步 $t$, $h_{t-1}$ 是模型能够看到的关于过去所有信息的（$x_{1,\cdots,t-1}$）的唯一通道
    - 无论过去的序列有多长，是10个词还是1000个词，它包含的所有信息都必须被压缩进 $h_{t-1}$ 这个固定大小的向量中
    - 计算上的梯度消失或爆炸问题
        - 在模型训练（反向传播）时，为了计算早期时间步的权重梯度，你需要将梯度沿着时间步一路乘回去。这个过程涉及到反复乘以同一个权重矩阵 $W_{hh}$
            - 取决于 $W_{hh}$ 的最大奇异值，如果 < 1，多次相乘 梯度会趋近于 0，如果 > 1，梯度会指数级增长，导致模型训练不稳定。
- context is lossy over moderate horizons
    - why? RNNs use fixed-size representations of the past.
- LSTM(2014) are ubiquitous across sequence-learning tasks： 引入了 $C_t$（cell state）和三个门控机制（输入门，遗忘门和输出门）
    - LSTM 最终传递给下一个时间步或者用于最终预测的，依然是一个固定大小的向量组合 $(h_t,C_t)$
- solution
    - Preserve the entire history, by using **attention**
    - 与其强迫模型将所有历史信息压缩成一个固定大小的向量，不如让模型在需要时，能够直接回顾和访问整个历史记录，并动态地决定哪些部分是重要的。
